```python
import pandas as pd
import pickle

data = pd.read_csv(
    "cdr_files_chosen_columns.csv",
    header=[0, 1],       
    low_memory=False     
)
```

```python
data.head(5)
```

```python
data.columns = data.columns.get_level_values(0)
```

horni uroven pro pojemenovani sloupcu

```python
data = data[[col for col in data.columns if "Unnamed" not in str(col)]]
```

zbavime se unnamed

```python
data.head(5)
```

```python
len(data.columns)
```

```python
for col in data.columns:
    print(col)
```

```python
data = data.astype({col: "string" for col in data.select_dtypes(include="object").columns})
```

```python
for col in ["dateTimeOrigination", "dateTimeConnect", "dateTimeDisconnect"]:
    data[col] = pd.to_numeric(data[col], errors="coerce")
    data[col] = pd.to_datetime(data[col], unit="s")
```
