Pocet telefonnich cislel koncicich '.0':

```python
end_0 = data['originalCalledPartyNumber'].str.endswith('.0', na=False)
pocet = end_0.sum()
print(f"Počet hodnot origcallednum končících na '.0': {pocet}")

end_0 = data['finalCalledPartyNumber'].str.endswith('.0', na=False)
pocet = end_0.sum()
print(f"Počet hodnot finalcallednum končících na '.0': {pocet}")

end_0 = data['callingPartyNumber'].str.endswith('.0', na=False)
pocet = end_0.sum()
print(f"Počet hodnot callingnum končících na '.0': {pocet}")
```

Ruzne zapisy volajicich cisel:

```python
start_plus = data['callingPartyNumber'].str.startswith('+', na=False)
pocet = start_plus.sum()
print(f"Počet hodnot callingnum zacinajicich na '+': {pocet}")

start_0 = data['callingPartyNumber'].str.startswith('0', na=False)
pocet = start_0.sum()
print(f"Počet hodnot callingnum zacinajicich na '0': {pocet}")

start_00 = data['callingPartyNumber'].str.startswith('00', na=False)
pocet = start_00.sum()
print(f"Počet hodnot callingnum zacinajicich na '00': {pocet}")

start_000 = data['callingPartyNumber'].str.startswith('000', na=False)
pocet = start_000.sum()
print(f"Počet hodnot callingnum zacinajicich na '000': {pocet}")
```
