```python
import pandas as pd
import pickle
data = pd.read_pickle("cdr_clean_unique_calls_NOb00.pkl")
```

```python
def uprav_cislo(number):
    number = str(number).strip()
    if number.startswith('+'):
        number = number[1:]
    if number.endswith('.0'):
        number = number[:-2]
    return number
```

```python
phone_columns = ['callingPartyNumber', 'originalCalledPartyNumber', 'finalCalledPartyNumber']

for column in phone_columns:
    data[column] = data[column].apply(uprav_cislo)
```

```python
for column in phone_columns:
    print(data[data[column].str.startswith('+')])
```

```python
for column in phone_columns:
    print(data[data[column].str.endswith('.0')])
```

```python
data.to_pickle('cdr_clean_unique_calls_update_num.pkl')
```
