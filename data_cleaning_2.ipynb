```python
phone_columns = ['callingPartyNumber', 'originalCalledPartyNumber', 'finalCalledPartyNumber']
summary_list = []
for column in phone_columns:
    lengths = data[column].astype(str).str.len()
    max_len = lengths.max()
    min_len = lengths.min()

    max_count = (lengths == max_len).sum()
    min_count = (lengths == min_len).sum()

    summary_list.append({'column': column,'column': column,'max_length': max_len,'max_count': max_count,'min_length': min_len,'min_count': min_count})


length_summary = pd.DataFrame(summary_list)
length_summary 
# minální a maximální délka tel, čísla a kolik jich je max_count
```

```python
#výpis čísle s max délkou
phone_columns = ['callingPartyNumber', 'originalCalledPartyNumber', 'finalCalledPartyNumber']

records = []

for col in phone_columns:
    col_str = data[col].astype(str)
    lengths = col_str.str.len()
    
    max_len = lengths.max()
    max_numbers = col_str[lengths == max_len]
    
    for num in max_numbers:
        records.append({'column': col, 'max_number': num})

max_numbers_df = pd.DataFrame(records)
max_numbers_df
```

```python
# kolik čísel s délkou větší než 15, 16, 17, 18, 19, 20, 25, 30, 35, 40
phone_columns = ['callingPartyNumber', 'originalCalledPartyNumber', 'finalCalledPartyNumber']

long_counts = {}

for col in phone_columns: 
    col_str = data[col].astype(str) 
    lengths = col_str.str.len()

    long_counts[col] = { 
        '>15': (lengths > 15).sum(),
        '>16': (lengths > 16).sum(), 
        '>17': (lengths > 17).sum(),
        '>18': (lengths > 18).sum(), 
        '>19': (lengths > 19).sum(),    
        '>20': (lengths > 20).sum(), 
        '>25': (lengths > 25).sum(), 
        '>30': (lengths > 30).sum(), 
        '>35': (lengths > 35).sum(), 
        '>40': (lengths > 40).sum(), 
    }
pd.DataFrame(long_counts)
```

```python
# příklady čísel s délkou větší než 15, 20, 25, 30, 35, 40
phone_columns = ['callingPartyNumber', 'originalCalledPartyNumber', 'finalCalledPartyNumber']

examples = {}

for col in phone_columns:
    col_str = data[col].astype(str)
    lengths = col_str.str.len()
    
    examples[col] = {
        '>15': col_str[lengths > 15].head(3).tolist(),
        '>20': col_str[lengths > 20].head(3).tolist(),
        '>25': col_str[lengths > 25].head(3).tolist(),
        '>30': col_str[lengths > 30].head(3).tolist(),
        '>35': col_str[lengths > 35].head(3).tolist(),
        '>40': col_str[lengths > 40].head(3).tolist(),
    }

pd.DataFrame(examples)
```
