Pocet unikatnich hovoru v ramci mesice:

```python
# vytvoreni sloupce 'month'
# dt je přístup k datetime atributům
data['month'] = data['dateTimeOrigination'].dt.to_period('M')

unique_month_calls = (
    data.drop_duplicates(['globalCallID_callId', 'origLegCallIdentifier', 
                          'month']).groupby('month').size()
)

print(unique_month_calls)
```

```python
data['week'] = data['dateTimeOrigination'].dt.to_period('W')

unique_week_calls = (
    data.drop_duplicates(['globalCallID_callId', 'origLegCallIdentifier', 
                          'week']).groupby('week').size()
)

print(unique_week_calls)
```
