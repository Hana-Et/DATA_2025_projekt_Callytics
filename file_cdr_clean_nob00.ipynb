```python
import pandas as pd
import pickle
data = pd.read_pickle("cdr_clean.pkl")
```

```python
start_b00 = data['callingPartyNumber'].str.startswith('b00', na=False)
pocet = start_b00.sum()
print(f"Počet hodnot zacinajicich na 'b00': {pocet}")
```

```python
data_no_b00 = data[~data['callingPartyNumber'].str.startswith('b00', na=False)]
data_no_b00.to_pickle('cdr_clean_NOb00.pkl')
```
