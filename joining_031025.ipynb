```python
import pandas as pd
import os
```

```python
DATA_PATH = r"C:\Users\adamk\Desktop\Digitální akademie DATA\Projekt\2024_01_03\2024_01_03" #r = row string a nepřepisuje zpětná lomítka
```

```python
filenames =os.listdir(DATA_PATH)  #vypíše všechny soubory uvnitř složky
```

```python
cesta_k_soboru = DATA_PATH + "\\" + filenames[0] #vytvoří cestu  k prvnímu souboru
```

```python
filenames[0]
```

```python
df = pd.read_csv(cesta_k_soboru, header=[0,1]) #header = záhlaví má 2 řádky
```

```python
filenames_cdr = []
filenames_cmr = []
for filename in filenames:
    if filename [:3] == "cdr":
        filenames_cdr.append(filename)
    if filename [:3] == "cmr":
        filenames_cmr.append(filename)

```

```python
cdr_dataframes = []
for filename in filenames_cdr: 
    file_path = DATA_PATH + "\\" + filename
    print(file_path)
    df = pd.read_csv(file_path, header=[0,1])
    cdr_dataframes.append(df)
    
```

```python
cdr_joined_df = pd.concat(cdr_dataframes) #tady se to musí opravit, pracujeme jen s 10, ale my chceme všechny; joined vyměnit za all
```

```python
cdr_joined_df.to_csv("cdr_soubory.csv") #uložení dat do souboru
```

```python
cmr_dataframes = []
for filename in filenames_cmr: 
    file_path = DATA_PATH + "\\" + filename
    print(file_path)
    df = pd.read_csv(file_path, header=[0,1])
    cmr_dataframes.append(df)
    
```

```python
cmr_joined_df = pd.concat(cmr_dataframes) #tady se to musí opravit, pracujeme jen s 10, ale my chceme všechny; joined vyměnit za all
```
