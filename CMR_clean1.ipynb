```python
for col in ['dateTimeStamp']:
    if col in data.columns:
        data[col] = pd.to_datetime(data[col], unit='s', errors='coerce')

print(data.head())
```

```python
print(" Min timestamp:", data['dateTimeStamp'].min())
print("Max timestamp:", data['dateTimeStamp'].max())
```

```python
null_counts = data.isnull().sum()
zero_counts = (data == 0).sum()
```

```python
summary = pd.DataFrame({
    'NULL_count': null_counts,
    'Zero_count': zero_counts
})
```

```python
summary['NULL_%'] = (summary['NULL_count'] / len(data) * 100).round(2)
summary['Zero_%'] = (summary['Zero_count'] / len(data) * 100).round(2)
```

```python
print(summary)
```

NULL hodnoty nahrazeny unkown v objects

```python
data['videoContentType'] = data['videoContentType'].fillna('unknown')
data['videoContentType_channel2'] = data['videoContentType_channel2'].fillna('unknown')
data['varVQMetrics'] = data['varVQMetrics'].fillna('unknown')
```

```python
print(data[['videoContentType', 'videoContentType_channel2','varVQMetrics']].isnull().sum())
```

```python
unknown_counts = (data[['videoContentType', 'videoContentType_channel2', 'varVQMetrics']] == 'unknown').sum()
print(unknown_counts)
```

prevedeni na Integer  directoryNum

```python
data['directoryNum'] = pd.to_numeric(data['directoryNum'], errors='coerce')
```

```python
data['directoryNum'] = data['directoryNum'].astype('Int64')
```

object na strig

```python
cols_to_convert = ['pkid', 'varVQMetrics', 'videoContentType', 'videoContentType_channel2']
data[cols_to_convert] = data[cols_to_convert].astype('string')
```

NULL hodnota v directoryNum nahrazena -1...symbolicka hodnota, znaci nezname cislo

```python
data['directoryNum'] = data['directoryNum'].fillna(-1)
```
