```python
import pandas as pd
import pickle
unique_calls = pd.read_pickle("cdr_clean_unique_calls_update_num2.pkl")
```

```python
columns = unique_calls[['unique_call_id', 'dateTimeConnect', 'dateTimeDisconnect']]
columns.head()
```

```python
# pro kazde unique_call_id najit min dateTimeConnect a maxTimeDisconnect
unique_calls = columns.groupby(['unique_call_id']).agg({"dateTimeConnect":"min","dateTimeDisconnect":"max"})
```

```python
unique_calls['duration_for_unique'] = unique_calls['dateTimeDisconnect'] - unique_calls['dateTimeConnect']

unique_calls.head()
```

```python
# nepocita s NaT, proto count je mensi nez celkovy pocet jedinecnych hovoru
unique_calls.describe()
```

```python
unique_calls.to_pickle('duration_unique_calls.pkl')
```

```python
unique_calls['month'] = unique_calls['dateTimeDisconnect'].dt.to_period('M')

unique_month_calls = (
    unique_calls.groupby('month').size()
)

print(unique_month_calls)
```

```python
unique_calls['week'] = unique_calls['dateTimeDisconnect'].dt.to_period('W')

unique_week_calls = (
    unique_calls.groupby('week').size()
)

print(unique_week_calls)
unique_week_calls.sum()
```

```python
unique_calls['week'] = unique_calls['dateTimeDisconnect'].dt.to_period('W')

unique_week_calls = (
    unique_calls.groupby('week').agg({'duration_for_unique':'median'}) #,'duration_for_unique':'count', 'duration_for_unique': 'max', 'duration_for_unique':'min'})
)

print(unique_week_calls)
unique_week_calls.sum() 
print()
```
