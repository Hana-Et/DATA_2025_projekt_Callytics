```python
#Čištění dat
import pandas as pd
import pickle
```

```python
data = pd.read_csv(r"C:\Users\adamk\Desktop\Digitální akademie DATA\Projekt\skript_02\cdr_files_chosen_columns.csv", header=[0, 1])
```

```python
#odstranění víceřádkého záhlaví
data.columns = data.columns.get_level_values(0)
```

```python
data.head()
```

```python
#odstranění sloupce s indexem, který pandas přidal při ukládání do CSV
data.drop(data.columns[0], axis=1, inplace=True)
```

```python
data.head()
```

```python
data.dtypes
```

```python
#převod telefonních čísel na string
for sloupec in ['callingPartyNumber', 'originalCalledPartyNumber', 'finalCalledPartyNumber']:
    data[sloupec] = data[sloupec].astype(str)
```

```python
# Oprava telefonních čísel - jsou různě dlouhá, nahrazujeme + za 00(Zbytek ignorujeme - telefonní čísla jsou různě dlouhá a v různých zápisech)
def uprav_cislo(number):
    number = str(number).strip()
    if number.startswith('+'):
        number = '00' + number[1:]
    return number
```

```python
phone_columns = ['callingPartyNumber', 'originalCalledPartyNumber', 'finalCalledPartyNumber']

for column in phone_columns:
    data[column] = data[column].apply(uprav_cislo)
```

```python
data.head()
```

```python
#Převod Unix timestamp sloupců na datetime
casove_sloupce = ['dateTimeOrigination', 'dateTimeConnect', 'dateTimeDisconnect']
for sloupec in casove_sloupce:
    data[sloupec] = pd.to_datetime(data[sloupec], unit='s')
```

```python
for sloupec in casove_sloupce:
    data[sloupec] = data[sloupec].replace(pd.Timestamp('1970-01-01 00:00:00'), pd.NaT) # NaT = Not a Time
print(data[casove_sloupce])
```

```python
#převod na stringy
for sloupec1 in ['callingPartyNumber', 'originalCalledPartyNumber', 'finalCalledPartyNumber']:
    data[sloupec1] = data[sloupec1].astype('string')


for sloupec2 in ['lastRedirectDn', 'pkid', 'origDeviceName', 'destDeviceName']:
    data[sloupec2] = data[sloupec2].astype('string')
```
