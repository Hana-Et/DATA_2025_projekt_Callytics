```python
import pandas as pd
```

```python
data = pd.read_csv('cdr_files_chosen_columns.csv', header=[0, 1]) 
```

```python
# divala jsem se, jakych nabyva hodnot sloupec globalCallID_callManagerId
data['globalCallID_callManagerId'].value_counts()
```

```python
# otazka na duplicity pomoci duplicated
duplicates = data[data['pkid'].duplicated(keep=False)]
duplicates
```

```python
#otazka na duplicity pomoci podminky a value_counts
id_counts = data['pkid'].value_counts()
id_counts[id_counts > 1]
```

```python
# zmena datoveho formatu utc na datetime 
data['dateTimeOrigination'] = pd.to_datetime(data['dateTimeOrigination']
                                             ['INTEGER'], unit='s', utc=True)
data['dateTimeConnect'] = pd.to_datetime(data['dateTimeConnect']['INTEGER'], 
                                         unit='s', utc=True)
data['dateTimeDisconnect'] = pd.to_datetime(data['dateTimeDisconnect']
                                            ['INTEGER'], unit='s', utc=True)
```

```python
# vytvoreni souboru se zmenenymi type
data.to_csv('cdr_files_chosen_columns_datetime.csv')
```
